```python
import pandas as pd
import numpy as np

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

from sklearn.ensemble import RandomForestClassifier
```

```python
'''
Predict if individual has diabetes (Outcome) based on provided features.
Use feature engineering, regularization and hyperparameter search to find best performing model.
At the very beggining, split the data into training and test dataset (70/30).

Pregnancies – Number of times pregnant.
Glucose – Plasma glucose concentration (mg/dL) from a 2-hour oral glucose tolerance test.
BloodPressure – Diastolic blood pressure (mm Hg).
SkinThickness – Triceps skinfold thickness (mm).
Insulin – 2-hour serum insulin level (μU/mL).
BMI – Body mass index (weight in kg / (height in m)²).
DiabetesPedigreeFunction – Diabetes likelihood based on family history.
Age – Age in years.

Outcome – Binary variable (1 = Diabetes, 0 = No Diabetes).

Dataset path: data/indian_diabetes.csv
'''

data = pd.read_csv('data/indian_diabetes.csv')

target_col = 'Outcome'
```

```python
data
```

```python
from sklearn.model_selection import train_test_split

X = data.drop(columns=[target_col])  
y = data[target_col] 

X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=1,
    stratify=y
)
```

```python
# Train random forest classifier
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=3
)
model.fit(X_train, y_train)

# Make predictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Print evaluation metrics
print(f"Train Accuracy: {accuracy_score(y_train, y_train_pred):.4f}")
print(f"Test Accuracy: {accuracy_score(y_test, y_test_pred):.4f}")
print(f"Train Precision: {precision_score(y_train, y_train_pred, average='weighted'):.4f}")
print(f"Test Precision: {precision_score(y_test, y_test_pred, average='weighted'):.4f}")
print(f"Train Recall: {recall_score(y_train, y_train_pred, average='weighted'):.4f}")
print(f"Test Recall: {recall_score(y_test, y_test_pred, average='weighted'):.4f}")
```
